# Game audio preview

Run from the repository root. The first section plays the five preprocessed
source recordings. The second compares the prepared channel levels for one sentence.

## Source recordings


```python
from IPython.display import Audio, display

from game.sentences import load_sentences


sentences = load_sentences()

for sentence in sentences:
    audio = sentence["audio"]
    fs = sentence["fs"]

    print(f'Sentence {sentence["id"]}: {sentence["text"]}')
    print(f"Duration: {len(audio) / fs:.2f} s | Sampling rate: {fs} Hz")

    display(Audio(audio, rate=fs, normalize=False))
```

## Processed levels

Change `SENTENCE_ID` to select a recording. Playback uses the gains stored in the
WAV files, with no extra normalization.


```python
from pathlib import Path
import soundfile as sf
from IPython.display import Markdown
from multiplayer.rules import LEVELS

SENTENCE_ID = 1
processed_dir = Path("game/assets/processed")

for suffix, label in [(f"ch_{level['channels']}", f"{level['channels']} channels")
                      for level in LEVELS] + [("original", "Original reference")]:
    path = processed_dir / f"sentence_{SENTENCE_ID:02d}_{suffix}.wav"
    signal, fs = sf.read(path)
    display(Markdown(f"**{label}**"))
    display(Audio(signal, rate=fs, normalize=False))
```
